```python
%matplotlib inline
import matplotlib.pyplot as plt
from spatialmath import SE2

THETA_DEG = 45          # <- change this, re-run, commit and push

T = SE2(0.2, 0.1, THETA_DEG, unit='deg')   # frame {A}: 0.2 m along x, 0.1 m along y, rotated THETA_DEG
print(T)
```

```python
SE2().plot(frame='0', color='k', width=0.5)
T.plot(frame='A', color='b')
plt.title(f'Frame {{A}} relative to {{0}}: {THETA_DEG} degrees')
plt.axis([-0.1, 0.4, -0.1, 0.4]);
```
